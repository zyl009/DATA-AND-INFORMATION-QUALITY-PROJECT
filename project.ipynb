###normalize the date

```python
# Function to normalize date
def normalize_date(value):
    # Dictionary for month corrections and seasons mapping
    month_corrections = {
        "Jut": "Jul",
        "rch": "Mar",
        "ept": "Sep",
        "prr": "Apr",
        }
    seasons_mapping = {
        "Winter": "12-01",
        "Spring": "03-01",
        "Summer": "06-01",
        "Fall": "09-01",
    }

    # Correct common typos
    for typo, correct in month_corrections.items():
        value = value.replace(typo, correct)

    # Patterns
    season_year_pattern = r'(Winter|Spring|Summer|Fall)[^\d]*(\d{4})'  # e.g., Winter 1942
    day_month_year_pattern = r'(\d{1,2})-([A-Za-z]{3})-(\d{4})'  # e.g., 19-Jul-1913
    month_year_pattern = r'([A-Za-z]{3})-(\d{4})'  # e.g., Aug-1987
    year_pattern = r'(\d{4})'  # e.g., 1901

    try:
       # Handle YYYY-MM-DD (already formatted)
        if re.match(r'^\d{4}-\d{2}-\d{2}$', value):
            return value
        # Check if the value matches the format 'YYYY.MM.DD'
        match = re.match(r'^(\d{4})\.(\d{2})\.(\d{2})$', value)
        if match:
            year, month, day = match.groups()
            # Replace invalid month/day components
            month = '01' if month == '00' else month
            day = '01' if day == '00' else day
            return f"{year}-{month}-{day}"  # Return in 'YYYY-MM-DD' format
        # Match season and year
        match = re.search(season_year_pattern, value, re.IGNORECASE)
        if match:
            season, year = match.groups()
            month_day = seasons_mapping[season.capitalize()]
            return f"{year}-{month_day}"
        # Handle phrases like 'mid-1870s'
        if "mid" in value:
            match = re.search(r'(\d{4})', value)
            if match:
                year = match.group(1)
                return f"{year}-06-01"  # Approximate mid-year date

        # Match day-month-year
        match = re.search(day_month_year_pattern, value)
        if match:
            day, month, year = match.groups()
            month_num = datetime.strptime(month, "%b").month
            return f"{year}-{month_num:02d}-{int(day):02d}"

        # Match month-year
        match = re.search(month_year_pattern, value)
        if match:
            month, year = match.groups()
            month_num = datetime.strptime(month, "%b").month
            return f"{year}-{month_num:02d}-01"

        # Match year only
        match = re.search(year_pattern, value)
        if match:
            year = match.group(1)
            return f"{year}-01-01"

    except Exception as e:
        # Log unexpected errors for debugging
        print(f"Error processing value: {value}. Error: {e}")

    # Return None for unprocessable rows
    return np.nan
```

```python
# Function to normalize time
def normalize_time(value):
    # Ensure the value is a string
    if not isinstance(value, str):
        return np.nan
    # Regex to find time already in HH:MM format
    match = re.search(r'\b(\d{1,2}):(\d{2})\b', value)
    if match:
        hour, minute = match.groups()
        return f"{int(hour):02d}:{int(minute):02d}"  # Return valid HH:MM format

    # Regex to find valid time patterns like hh:mm or hhm
    match = re.search(r'(\d{1,2})h(\d{2})', value)
    if match:
        hour, minute = match.groups()
        return f"{int(hour):02d}:{int(minute):02d}"

    # Map descriptions to approximate times
    time_map = {
        'early morning': '06:00',
        'late morning': '10:30',
        'morning': '08:00',
        'early afternoon': '13:00',
        'late afternoon': '16:30',
        'afternoon': '15:00',
        'late evening': '21:00',
        'evening': '19:00',
        'dusk': '18:30',
        'night': '22:00',
        'noon': '12:00',
        'midnight': '00:00'
      }
    for key, mapped_time in time_map.items():
        if key in value.lower():
            return mapped_time


    # Return None for unprocessable rows
    return np.nan


# Sample data
sampledTime = pd.DataFrame({
    'Time': ['08h00', 'wwwda15h00bdbgrfd', '15h45lll', 'Dusk','dusklllll', 'morning', 'MORNING','late late Late morning', 'siskks early afternoon', 'kcdoak night 6154', 'unknown']
})

# Apply normalization
sampledTime['Normalized Time'] = sampledTime['Time'].apply(normalize_time)
sampledTime
```

```python
indexer = recordlinkage.index.SortedNeighbourhood(
        on='href', window=9)

candidate_links = indexer.index(data)
candidate_links
```

```python
data.columns
```

```python
compare_cl = recordlinkage.Compare()

compare_cl.exact('Case Number', 'Case Number', label='Case Number')
compare_cl.exact('Date', 'Date', label='Date')
compare_cl.exact('Time', 'Time', label='Time')
compare_cl.exact('Year', 'Year', label='Year')
compare_cl.exact('Type', 'Type', label='Type')
compare_cl.exact('Activity', 'Activity', label='Activity')
compare_cl.exact('Activity Detail', 'Activity Detail', label='Activity Detail')
compare_cl.exact('Age', 'Age', label='Age')
compare_cl.exact('Fatal', 'Fatal', label='Fatal')

compare_cl.string('Country', 'Country', method='jarowinkler', threshold=0.9, label='Country')
compare_cl.string('Area', 'Area', method='jarowinkler', threshold=0.9, label='Area')
compare_cl.string('Location', 'Location', method='jarowinkler', threshold=0.9, label='Location')
compare_cl.string('Name', 'Name', method='jarowinkler', threshold=0.9, label='Name')
compare_cl.string('Injury', 'Injury', method='jarowinkler', threshold=0.9, label='Injury')
```

```python
features = compare_cl.compute(candidate_links, data)
matches = features[features.sum(axis=1) > 11]
matches
```

```python
# Drop duplicates (e.g., keep the first occurrence)
duplicate_indices = matches.index.get_level_values(1)
data= data.drop(duplicate_indices)
```
